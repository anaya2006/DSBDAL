A7 Text Analytics 1. Extract Sample document and apply following document preprocessing methods: Tokenization, POS Tagging, stop words removal, Stemming and Lemmatization. 2. Create representation of document by calculating Term Frequency and Inverse Document Frequency

```python
import nltk
```

```python
document = """Artificial Intelligence is transforming modern industries.
Many companies use machine learning algorithms to analyze customer behavior,
predict future trends, and automate repetitive tasks.

Data scientists collect large volumes of structured and unstructured data
from websites, mobile applications, and social media platforms.

Natural Language Processing helps computers understand human language
and improve chatbot interactions.

However, organizations must also address privacy, security,
and ethical concerns while developing intelligent systems."""

print("Sample document \n", document)
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk.tokenize import word_tokenize

tokens= word_tokenize(document)

print("Tokens: \n" , tokens)
```

```python
from nltk import pos_tag 
nltk.download('universal_tagset')
pos_tags= pos_tag(tokens, tagset='universal')

print("POS tagging : \n", pos_tags)
```

```python
from nltk.corpus import stopwords 

stop_words = set(stopwords.words('english'))

filtered_tokens = []

for word in tokens:
    if word.lower() not in stop_words:
        filtered_tokens.append(word)

print("Tokens after Stopward Removal: \n", filtered_tokens)
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

stemmed_words = []

for word in filtered_tokens:
    stemmed_words.append(stemmer.stem(word))

print("Stemmed words: \n", stemmed_words)
```

```python
from nltk.stem import WordNetLemmatizer 

lemmatizer = WordNetLemmatizer()

lemmatized_words =[]

for word in filtered_tokens:
    lemmatized_words.append(lemmatizer.lemmatize(word))

print("Lemmatized Words: \n", lemmatized_words)
```

```python
from collections import Counter

word_count = Counter(lemmatized_words)

total_words = len(lemmatized_words)

tf={}

for word, count in word_count.items():
    tf[word] = count/total_words

print("\n TF : \n")

for word,value in tf.items():
    print(f"{word} : {value:.4f}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer 

documents = [
    "Artificial Intelligence is transforming industries using machine learning.",
    
    "Data scientists analyze structured and unstructured data.",
    
    "Natural Language Processing improves chatbot interactions.",
    
    "Organizations must address privacy and security concerns."
]

vectorizer = TfidfVectorizer()

X= vectorizer.fit_transform(documents)

feature_names = vectorizer.get_feature_names_out()

print("\n\n TF-IDF Values : ")

for word,score in zip(feature_names , X.toarray()[0]):
    print(f"{word} : {score:.4f}")
```
