```python
import pandas as pd
import numpy as np
np.random.seed(50)

data= {
  'name': ['student_' + str(i) for i in range (1,51) ],
  'roll' : range(1,51),
    'age': np.random.randint(18,25,size=50),
  'attendance' : np.random.choice((0,100),size=50),
   'gender': np.random.choice(('Male', 'Female', ' '), size=50),
    'grade' : np.random.choice(('O','A','B','C','F',' '),size=50),
    'score' : [np.random.randint(50, 100, size=3).tolist() for _ in range(50)],
}

df=pd.DataFrame(data)
df.head()
```
